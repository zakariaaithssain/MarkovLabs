```python
# 5. 
n = P.shape[0]

A = P.T - np.eye(n)
A[-1, :] = 1

b = np.zeros(n)
b[-1] = 1

steady_state = np.linalg.solve(A, b)
pd.Series(steady_state.round(2), index =[i for i in range(1,8)] + [9]).astype(int)
```

```python
# 6. 

new_order = [6, 7, 0, 1, 2, 3, 4, 5]  
P_canonical = P[np.ix_(new_order, new_order)]

state_labels = [7, 9, 1, 2, 3, 4, 5, 6]
print("canonical form:\n", pd.DataFrame(P_canonical, index=state_labels, columns=state_labels), end="\n\n")

Q = P_canonical[2:, 2:]
print("Q:\n", pd.DataFrame(Q, index=[1,2,3,4,5,6], columns=[1,2,3,4,5,6]), end = "\n\n")


N = np.linalg.inv(np.eye(Q.shape[0]) - Q).round(2)

print("fondamental matrix: \n", pd.DataFrame(N, index = range(1, 7), columns = range(1, 7)))
```

```python
v_souris={1:[2,4],2:[1,3,5],3:[2,6],4:[1,5,7],5:[2,4,6],6:[3,5,9],7:[4],9:[9]}
v_chat={1:[1,4,1,2],2:[2,5,1,3],3:[3,6,2,3],4:[1,7,4,5],5:[2,5,4,6],6:[3,6,5,9],7:[4,7,7,7],9:[6,9,9,9]}

acces=[1,2,3,4,5,6,7,9]
stts=[(s,c) for s in acces for c in acces]
n=len(stts)

P=np.zeros((n,n))
for i in range(n):
    s=stts[i][0]
    c=stts[i][1]
    if s==c or s==9:
        P[i,i]=1.0
    else:
        ms=v_souris[s]
        mc=v_chat[c]
        for s2 in ms:
            for c2 in mc:
                j=stts.index((s2,c2))
                P[i,j]+=(1/len(ms))*(1/4)

pi0=np.zeros(n)
pi0[stts.index((1,7))]=0.5
pi0[stts.index((3,7))]=0.5

pi5=pi0@P@P@P@P@P

p_fromage=0
p_mange=0
for i in range(n):
    s=stts[i][0]
    c=stts[i][1]
    if s==9 and c!=9:
        p_fromage+=pi5[i]
    if s==c:
        p_mange+=pi5[i]

print("P(fromage apres 5 etapes) =",round(p_fromage,4))
print("P(mangee apres 5 etapes)  =",round(p_mange,4))

trans=[]
for i in range(n):
    s=stts[i][0]
    c=stts[i][1]
    if s!=c and s!=9:
        trans.append((s,c))
nt=len(trans)
#On applique l'algo d'itératoin de valeur pour trouver la politique optimale'
V=np.zeros(nt)
for iteration in range(1000):
    Vn=np.zeros(nt)
    for i in range(nt):
        s=trans[i][0]
        c=trans[i][1]
        best=999
        for c2 in set(v_chat[c]):
            val=0
            for s2 in v_souris[s]:
                p=1/len(v_souris[s])
                if s2==c2:
                    val=val + p*1
                elif s2==9:
                    val=val + p*999
                else:
                    j=trans.index((s2,c2))
                    val=val + p*(1 + V[j])
            if val < best:
                best=val
        Vn[i]=best
    if np.max(np.abs(Vn - V)) < 1e-6:
        break
    V=Vn

print("\npolitique optimale du chat :")
for i in range(nt):
    s=trans[i][0]
    c=trans[i][1]
    best_c2=c
    best_val=999
    for c2 in set(v_chat[c]):
        val=0
        for s2 in v_souris[s]:
            p=1/len(v_souris[s])
            if s2==c2:
                val=val + p*1
            elif s2==9:
                val=val + p*999
            else:
                j=trans.index((s2,c2))
                val=val + p*(1 + V[j])
        if val < best_val:
            best_val=val
            best_c2=c2
    print("souris =",s,"chat =",c,"chat va en",best_c2)

W=np.zeros(nt)
for iteration in range(1000):
    Wn=np.zeros(nt)
    for i in range(nt):
        s=trans[i][0]
        c=trans[i][1]
        best=-999
        for s2 in v_souris[s]:
            val=0
            for c2 in v_chat[c]:
                p=1/4
                if s2==9 and c2!=9:
                    val=val + p*999
                elif s2==c2:
                    val=val + p*(-999)
                else:
                    j=trans.index((s2,c2))
                    val=val + p*W[j]
            if val > best:
                best=val
        Wn[i]=best
    if np.max(np.abs(Wn - W)) < 1e-6:
        break
    W=Wn

print("\npolitique optimale de la souris :")
for i in range(nt):
    s=trans[i][0]
    c=trans[i][1]
    best_s2=s
    best_val=-999
    for s2 in v_souris[s]:
        val=0
        for c2 in v_chat[c]:
            p=1/4
            if s2==9 and c2!=9:
                val=val + p*999
            elif s2==c2:
                val=val + p*(-999)
            else:
                j=trans.index((s2,c2))
                val=val + p*W[j]
        if val > best_val:
            best_val=val
            best_s2=s2
    print("souris =",s,"chat =",c,"souris va en",best_s2)
```
